#### 1. James just concluded 5 maths tests. Here are his scores: 78, 90, 67, 52, 84
a. Store his scores in a list.      
b. What are his highest and lowest scores?     
c. Find the average of his scores (Hint: Use sum() and len())     
d. Print a message: “James’s best score is 90, and his lowest score is 52. His average score is 74.2”



```python
james_score = [78,90,67,52,84]

best_score = max(james_score)
lowest_score = min(james_score)

average_score = sum(james_score)/len(james_score)

print(f'“James’s  best score is {best_score}, and his lowest score is {lowest_score}.\
 His average score is {average_score}”')
```

#### 2. Alice went to the market to get groceries. Here is a list of the amount she spent on 7 items; 1200, 1000, 450, 780, 870, 1100, 1500.
a. Store the amounts in a list.  
b. What is the most expensive amount?  
c. What is the least expensive amount?  
d. Get the three most expensive amounts using slicing. 

```python
groceries_items = [1200, 1000, 450, 780, 870, 1100, 1500]

most_expensive = max(groceries_items)
least_expensive = min(groceries_items)

three_most_expensive = sorted(groceries_items, reverse=True)[:3]

print(f'The most expensive amount is {most_expensive}.\
        \nThe least expensive amount is {least_expensive}.\
        \nThe three most expensive amounts are {three_most_expensive}')
```
